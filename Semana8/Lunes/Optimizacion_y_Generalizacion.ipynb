```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
#el cross_val_score es el que permite evaluar el modelo varias veces utilizando vairas partes del modeloo del conjunto de datos
#el KFold nos ayuda a dividir los datos en pequeños bloques                     
from sklearn.model_selection import cross_val_score,KFold

print("Descenso del Gradiente y validacion Cruzada KFold")

#simulacion del descenso del gradiente 
#funcion que ira simulando como el descenso del gradiente va ir modificando un peso para ir intentando acercarlo al valor optimo, 
# lo que representa el error minimo
#alpha es la tasa de aprendizaje, cuantos saltos va a estar dando   

def simular_gradiente(alpha,iteraciones=10):
    
    w=10.0#ayudar a calcular el gradiente,es decir como me va a ir aumentando y disminuyendo el valor del gradiente a medida nos vamos moviendo de un punto a otro
    print(f"\nProbando Tasa de Aprendizaje= {alpha}: ")
    for i in range(1,iteraciones+1):
        
        gradiente=2*w
        #actualizar el peso
        w=w-alpha*gradiente#formula
        #comprueba si se calcula de w se salio de control, o si se convirtiio en nan(valor numerico invalido) se verifica si el valor de abosoluto de  w supero los 100,000=1e5
        if(np.isnan(w)or abs(w)>1e5):
            
            print(f"Iteacion {i}: Divergencia! El modelo exploto")
            break
        print(f"Iteracion {i}: Peso w= {w:.4f}:")

#acercando hacia el error minimo
simular_gradiente(alpha=0.1)
simular_gradiente(alpha=1.1)


#validacion cruzada (KFold)
np.random.seed(42)
#genera una matriz de numeros aleatorios, genera 100 filas con 3 columnas y le va a poner valores decimales que van a estar entre 0 y 1, sin incluir el 1
#ejem: 0.2 0.3 0.8
#      0.1 0.6 0.3
X=np.random.rand(100,3)            
#                                   media=0,   std=0.5,  100 valores aleatorios con distribucion normal
y=3*X[:,0]+2*X[:,1]+np.random.normal(0,0.5,100)
kf=KFold(n_splits=5,shuffle=True,random_state=42)
#aqui hace el entrenamiento 
scores=cross_val_score(
    
    LinearRegression(),#crea el nuevo modelo de regresion lineal
    X,#variabel predictora
    y,#vaiarable objetivo
    cv=kf,# 5 divisiones de los datos
    scoring='neg_mean_squared_error'#indica que modelo sera evaluado 
    
)
mse_scores=-scores

print(f"Resultados de los 5 Folds (MSE por pliegue): {np.round(mse_scores,4)}")
print()        
print(f"MSE Promedio Estimado de Generalizacion: {mse_scores.mean():.4f} (± {mse_scores.std():.4f})")

```
