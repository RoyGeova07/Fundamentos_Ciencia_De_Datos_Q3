```python
import pandas as pd
import requests


#url de la api - sin autorizacion
url_api="https://jsonplaceholder.typicode.com/users"

#realizar peticion
headers={'Users-Agent':'Roy Umaña - Unitec' }
respuesta=requests.get(url_api,headers=headers)

#verificar respuesta
if respuesta.status_code==200:
    print("Conexion exitosa con la API")
    #convertir respuesta en el json
    dato_json=respuesta.json()#permite la respuesta convertir en JSON
    
    #extraer las variable
    datos=[]
    for usuario in dato_json:
        datos.append({
            
            "Nombre":usuario["name"],
            "Correo":usuario["email"],
            "Ciudad":usuario["address"]["city"],
            "Empresa":usuario["company"]["name"],
            "Latitud":usuario["address"]["geo"]["lat"],
            
        })
        
    #estructurar la informacion del dataframe
    df=pd.DataFrame(datos)
    #print(df)

    #creacion del archivo csv
    df.to_csv("Usuarios.csv",index=False)
    print("\nArchivo creado exitosamente")
    
else:
    print("Error: ",respuesta.status_code)

```
