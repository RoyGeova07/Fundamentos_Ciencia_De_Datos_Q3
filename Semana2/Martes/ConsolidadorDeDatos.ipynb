```python
import pandas as pd
print("Consolidador de Datos")
#Datos obtenidos desde un scraping web
#Este DataFrame contiene informacion de paises y su PIB.
df_scraping=pd.DataFrame({
    'Pais':['Estados Unidos','China','Alemania','Japon','Honduras','Guatemala'],
    'PIB':[28.7,18.5,4.5,4.1,0.035,0.4]  
})
#Datos simulados de una API externa
#Este DataFrame agrega el codigo del pais y un indice de conectividad.
df_api=pd.DataFrame({
    'Pais':['Honduras','Alemania','Estados Unidos','China','Japon'],
    'Codigo_pais':['HN','DE','US','CN','JP'],
    'Indice_conectividad':[42.1,88.5,92.3,85.1,89.8]  
})
#Consolidacion de los datos usando una union (merge) por la columna 'Pais'
#how='right' mantiene todas las filas de df_api (el DataFrame derecho)
#y agrega las columnas de df_scraping solo cuando el pais aparece en ambos DataFrames.
df_combinado=pd.merge(df_scraping,df_api,on='Pais',how='right')
print(df_combinado)
#Verificacion de integridad
print("\nVerificación de integridad")
print(f"Total de registros cargados: {df_combinado.shape[0]}")
print("Tipos de datos detectados")
print(df_combinado.dtypes)
#Persistencia: exportar el DataFrame consolidado a archivos
print("\n")
df_combinado.to_csv("combinado.csv", index=False)
#Guardar tambien en formato Parquet, util para analisis y almacenamiento eficiente.
df_combinado.to_parquet("combinados.parquet", index=False)
print("Archivos generados")
```
