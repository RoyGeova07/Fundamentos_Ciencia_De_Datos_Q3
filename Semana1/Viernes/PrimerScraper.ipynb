```python
#importar librerias

import requests
from bs4 import BeautifulSoup
import pandas as pd
import time


#definir URL
url="https://books.toscrape.com"

#enviar solicitud http

headers={'User-Agent':'Roy Umaña - roy_umana@unitec.edu' }

#peticion, agarra el html crudo
response=requests.get(url,headers=headers)#trae todo el codigo html

#tiempo de espera entre la solcitud 
time.sleep(2)

#verificar la respuesta
if response.status_code==200:
    print("Pagina obtenida correctamente")
else:#               error de codigo
    print("Error: ",response.status_code)
    
    
#analisis del html, parsear html
soup=BeautifulSoup(response.text,"html.parser")#contenido a parsear y el formato q utilizara
#print(soup)
libros=soup.find_all("article",class_="product_pod")
#lista de libros que estan dentro de las etiquetas
#print(libros)

#.strip() - recupera el texto y elimina los saltos de linea osea los espacios

#crear lista vacia, donde se almacenar los libros
datos=[]
#   lista
for libro in libros:
    titulo=libro.h3.a["title"]
        # etiqueta y clase 
    precio=libro.find("p",class_="price_color").text
    disponibilidad=libro.find("p",class_="instock availability").text.strip()
    datos.append([titulo,precio,disponibilidad])
    
#guardar la lista  en un dataframe
df=pd.DataFrame(datos,columns=["Titulo","Precio","Disponibilidad"]) 



#crear el archivo csv
df.to_csv("libros.csv",index=False)


#visualizar el dataFrame
print(df.head())
#df.head()

print("\nScraping finalizado")
```
