```python
import pandas as pd
import numpy as np
import re#permite modulo de pyhton que permite trabajar con expresiones regulares para los correos electronicos

#dataset sucio
datos_sucios={
    
    'Usuario_id':[101,102,103,104,102],
    'Correo':['carlosgmail.com','ana#gmail.com',None,'maria@domain.com','ana#gmail.com'],
    'Edad':[28,-5,34,150,-5],
    'Fecha_registro':['2024-01-15','2024-02-01','2026-12-31','2024-03-10','2024-02-01'],
    'Estado_civil':['Soltero','Casado','Soltero','Casado','Casado'],
    'Años_casado':[0,5,12,0,5]
    
}

df_calidad=pd.DataFrame(datos_sucios)
print("Dataset con datos sucion")
print(df_calidad)

#evaluar dimensiones de calidad
total_filas=len(df_calidad)

#completitud (porcentaje de celdas no nulas en el correo)
completitud_correo=(df_calidad['Correo'].notnull().sum()/total_filas)*100

#unicidad (el porcentaje de registros no duplicados)
unicidad=(len(df_calidad['Usuario_id'].drop_duplicates())/total_filas)*100


#validez (formato de correo /rangos de edad)
#r: cadena cruda - ^: inicio -  \w\: caracteres alpha numericos - +@
patron_email=r'^[\w\.-]+@[\w\.-]+\.\w+$'
#seleciona correos, aplica una funcion a cada correo, lambda x: recibe cada valor q tiene la columna de correo, re.match(si el correo coincide con el patron que saco con la expresiones regulares, devuelcve true si coincide y false si no)
#en si me dara los correos que son validos
correos_validos=df_calidad['Correo'].apply(lambda x:bool(re.match(patron_email,str(x)))).sum()
validez_correo=(correos_validos/total_filas)*100#porcentaje de validez de los correos

#Edades validas
edades_validas=df_calidad['Edad'].apply(lambda x:0<=x<=100).sum()
validez_edad=(edades_validas/total_filas)*100

#consistencia (soltero vs anios casado)
inconsistencias=df_calidad[(df_calidad['Estado_civil']=='Soltero')&(df_calidad['Años_casado']>0)]
consistencia=((total_filas-len(inconsistencias))/total_filas)*100

#matriz de calidad
print()
print("====Matriz de calidad====")
print(f"Completitud (campo correo): {completitud_correo:.1f}%")
print(f"Unicidad (registros unicos): {unicidad:.1f}%")
print(f"Validez (formato de correo): {validez_correo:.1f}%")
print(f"Validez (rango de edad): {validez_edad:.1f}%")
print(f"Consistencia (estado civil vs años casado): {consistencia:.1f}%")
```
