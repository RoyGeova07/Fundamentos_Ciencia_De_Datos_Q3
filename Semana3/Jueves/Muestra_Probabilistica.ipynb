```python


import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split#sirve para enteder modelos de inteligencia artificial

#simulador de muestreo
print("Simulador de muestreo")
print()

#generar la poblacion
np.random.seed(42)
poblacion=pd.DataFrame({
    
    'Ciente_id':range(1,10001),
    #                                                             probabilidad - 25% femeninos y 75% masculinos
    'Genero':np.random.choice(['Femenino','Masculino'],size=10000,p=[0.25,0.75]),
    # loc da la media
    'Ingresos':np.random.normal(loc=2600,scale=550,size=10000)
    
    
})

print(poblacion)
print()

print("Calculo de los parametros poblacionales reales")
print()

media_poblacional=poblacion['Ingresos'].mean()
#                              normalize devuelve las proporciones que serua el total de la categoria por el total de los registros 
proporcion_mujeres=poblacion['Genero'].value_counts(normalize=True)['Femenino']

print(f"Media real de los ingresos es: {media_poblacional:.2f}") #media_poblacional:2.f
print(f"Proporcion real de mujeres es: {proporcion_mujeres:.2f}")
print()


print("Extraccion de las muestras")
print()

n_objetivo=400
#muetreo aleatorio simple
#sample lo que hace es tomar una muestra de los datos del dataframe de poblacion
#                 
muestra_as=poblacion.sample(n=n_objetivo,random_state=42)

#muestreo sistematico
# se usa '//' para que el resultado salga entero
#lo que hace el k es dice: 'voy a seleccionar una persona cada 25 personas'
k=len(poblacion)//n_objetivo
#el iloc es index location, sirve para acceder a filas por su posicion, no por su nombre 
# el .head() hace se quede unicamente con las primeras 400 filas 
muestra_sis=poblacion.iloc[::k].head(n_objetivo)


#muestreo estratificado
#nomneclatura que ese valor hay que ignorarlo cuando se usa '_'                            buscar que hace el random_state
muestra_estr,_=train_test_split(poblacion,train_size=n_objetivo,stratify=poblacion['Genero'],random_state=42)


print()
print("Estadisticos muestrales obtenidos")
print()

def evaluar_muestra(nombre,df_muestra):
    media_m=df_muestra['Ingresos'].mean()
    #obtiene la proporcion de solo el geneero femenino
    prop_m=df_muestra['Genero'].value_counts(normalize=True).get('Femenino',0)*100
    error_media=abs(media_poblacional-media_m)
    print(f"[{nombre}] -> Media:{media_m:.2f}, Error:{error_media:.2f}, Porcentaje de Mujeres:{prop_m:.2f}%")# agregar .2f, sale error al agregarlo
    
evaluar_muestra("Muestreo aleatorio simple",muestra_as)
evaluar_muestra("Muestreo Sistematico",muestra_sis)
evaluar_muestra("Muestreo Estratificado",muestra_estr)

    
    
```
