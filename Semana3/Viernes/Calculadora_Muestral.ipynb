```python


import math

print("Calculadora de diseño muestral")
print()

#calcular funcion para poder sacar muestras de poblacion infinitas
def calcular_muestra_pob_infinita(confianza,margen_error,p=0.5):
    
    #diccionario
    valores_z={0.90:1.645,0.95:1.96,0.99:2.57,}
    z=valores_z[confianza]
    q=1-p
    n=(z**2*p*q)/(margen_error**2)
    return math.ceil(n)#devuelve el valor de n redondeeado hacia arriba, osea redondearme al entero mas cercano

#tamanio de muestra cuando yo tengo la poblacion q es finita
def calcular_muestra_pob_finita(N,confianza,margen_error,p=0.5):#n es la poblacion
    
    valores_z={0.90:1.645,0.95:1.96,0.99:2.57,}
    z=valores_z[confianza]
    q=1-p
    # numerador     -       denominador
    n=(N*(z**2)*p*q)/((margen_error**2)*(N-1)+(z**2)*p*q)
    return math.ceil(n)

finita=calcular_muestra_pob_finita(5000,0.95,0.05)
print(finita)


#datos de prueba
usuarios_activos=50000
error_tolerado=0.03
confianza_deseada=0.95

n_infinita=calcular_muestra_pob_infinita(confianza_deseada,error_tolerado)
n_finita=calcular_muestra_pob_finita(usuarios_activos,confianza_deseada,error_tolerado)

print("Resultado del diseño de muestra")
print()
print(f"Poblacion Infinita -> Tamaño de muestra: {n_finita} usuarios")
print(f"Poblacion Finita -> Tamaño de muestra: {n_infinita} usuarios (de {usuarios_activos})")

print()
print("Analisis de sensibilidad")
print()

for err in [0.05,0.03,0.01]:
    resultado=calcular_muestra_pob_infinita(0.95,err)
    print(f"Con un nivel de confianza del 95% y un error del {err*100:.0f}% -> se necesitan {resultado} usuarios de muestra")
```
