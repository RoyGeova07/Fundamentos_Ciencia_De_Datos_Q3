```python


import pandas as pd
import numpy as np

print("Pipeline de ingenieria de caracteristicas")

#crear dataset de clientes que tengas variables nominales y categoricas
datos_streaming={
    #                    3 digitos, 6 clientes 
    'Cliente_id':[f"USR_{i:03d}" for i in range(1,7)],
    'Pais':['Honduras','España','Honduras','Japon','España','Japon'],#variable nominal, (no tiene orden)
    'Plan_suscripcion':['Basico','Premium','Basico','Familiar','Premium','Familiar'],#variable ordinal (si tiene jerarquia)
    'Fecha_registro':['2023-01-15','2024-05-22','2023-11-02','2025-08-14','2024-12-01','2026-03-10']# fecha como texto
    
}


df_streaming=pd.DataFrame(datos_streaming)
print()
print("DATOS ORIGINALES")
print(df_streaming)


print()
print("LABEL ENCONDING -- Variables ordinales")

#crear copia del datframe
df_encoding=pd.DataFrame(df_streaming)
#definir orden jerarquico de los planes
#Basico < Premium < Familiar
mapeo_planes={'Basico':0,'Premium':1,'Familiar':2}
# map() reemplaza cada texto por el numero indicado
# en el diccionario mapeo_planes.
# Ejemplo:
# Basico->0
# Premium->1
# Familiar->2
# Resultado:
# Basico
# Premium
# Basico
# se convierte en:
# 0
# 1
# 0
df_encoding['Plan_encoded']=df_streaming['Plan_suscripcion'].map(mapeo_planes)
print(df_encoding)

print()
print("ONE HOT ENCONDING -- Variables nominales")


#get_dummies() convierte una variable categorica
#en varias columnas binarias (0 y 1).
# Pais
# Honduras
# España
# Japón
# se convierte en
# Pais_Honduras  Pais_España  Pais_Japon
#      1               0            0
#      0               1            0
#      0               0            1
# prefix="Pais"
# agrega la palabra "Pais_" delante del nombre
# de cada nueva columna.
# dtype=int
# hace que los valores sean enteros (0 y 1)
# en lugar de True y False.
df_dummies=pd.get_dummies(df_streaming['Pais'],prefix="Pais",dtype=int)
#axis significa el eje sobre el que se hara la union,axis=0  -> une filas (verticalmente)
#Ejemplo: AB + CD = ABCD, 
# axis=1 -> une columnas (horizontalmente)
#Ejemplo: 
# Nombre      Edad
# Ana          20
# Luis         25
# +
# País
# Honduras
# España
# =
# Nombre   Edad   País
# Ana       20    Honduras
# Luis      25    España
#Aqu se usa axis=1 porque queremos agregar las nuevas columnas de los paises al mismo cliente.
df_transformado=pd.concat([df_encoding,df_dummies],axis=1)#unir horizontalmente las columnas de los dataframes
print(df_transformado)

print()
print("INGENIERIA DE CARACTERISTICAS")
# convertir columna de texto plano a tipo fecha
df_transformado['Fecha_registro']=pd.to_datetime(df_transformado['Fecha_registro'])

#extraer las nuevas variables a partir de la fecha
df_transformado['anio']=df_transformado['Fecha_registro'].dt.year
df_transformado['mes']=df_transformado['Fecha_registro'].dt.month
df_transformado['dia']=df_transformado['Fecha_registro'].dt.dayofweek


print()
print(df_transformado)

print()
print("RESULTADO FINAL PARA NUESTRO MODELO DE IA")
columnas_finales=['Plan_encoded','Pais_España','Pais_Honduras','Pais_Japon','anio','mes','dia']
print(df_transformado[columnas_finales])
```
