## Esto se basa en el impacto economico que puede tener un modelo de mantenimiento predictivo en una fabrica 

```python
print("Traductor Financiero de Modelos de IA")
#Primero se establece los costos. Y se  revisa una maaquina si se daño  o no etc
costo_reparacion_preventiva=500 #lps
#costo VP/FP
costo_catastrofe_paro=10000
#costo FN


#Resultados del modelo 1000 maquinas
vp=85#fallas detectada a tiempo
fn=15#fallas no detectadas a tiempo
fp=30#falsas alarmas
vn=870#maquina funcionaron normalmente

#Calculo del impacto real financiero
ahorro_fallas_prevenidas=vp*(costo_catastrofe_paro-costo_reparacion_preventiva)
#cuanto cuesta las falsa
costo_falsas_alarmas=fp*costo_reparacion_preventiva
perdida_total_fallas_pasadas=fn*costo_catastrofe_paro
beneficio_neto_modelo=ahorro_fallas_prevenidas-costo_falsas_alarmas

#reporte del impacto financiero
print("Reporte de Impacto Financiero para la Gerencia")
print()
print(f"Ahorro por fallas capturadas a tiempo (Sensibilidad (Recall) {(vp/(vp+fn))*100}%): Lps {ahorro_fallas_prevenidas:,.2f}")
print(f"Costo operativo por falsas alarmas (FP): Lps {costo_falsas_alarmas:,.2f}")
print(f"Perdida por fallas no detectadas (FN): Lps {perdida_total_fallas_pasadas:,.2f}")
print("----------------------------------------")
print()
print(f"RETORNO NETO DEL MODELO DE IA: Lps {beneficio_neto_modelo:,.2f}")



```
