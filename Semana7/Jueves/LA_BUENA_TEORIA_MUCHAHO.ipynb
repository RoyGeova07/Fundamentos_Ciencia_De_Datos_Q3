# Umbral y Matriz de Confusión

## 1. Umbral

El **umbral** es el límite o la línea que uno decide establecer para separar una clase de otra.

La mayoría de los modelos de clasificación no dicen directamente:

> "Esto es spam".

En su lugar, calculan una **probabilidad**. Por ejemplo:

> "Hay un 75% de probabilidad de que este correo sea spam".

Por defecto, el umbral casi siempre es **0.50 (50%)**.

- Si la probabilidad es **≥ 50%**, el modelo clasifica como **Positivo**.
- Si la probabilidad es **< 50%**, el modelo clasifica como **Negativo**.

### ¿Se puede cambiar el umbral?

Sí. El umbral se puede modificar dependiendo de lo que queramos priorizar.

- **Umbral alto (80%)** → El modelo será más **estricto** para marcar algo como positivo.
- **Umbral bajo (20%)** → El modelo será más **flexible** para marcar algo como positivo.

---

# 2. Matriz de Confusión

Para entender la matriz de confusión, imaginemos un modelo que detecta si una transacción bancaria es:

- **Fraude → Positivo**
- **Transacción normal → Negativo**

|                                      | Realidad: Es Fraude (Positivo) | Realidad: No es Fraude (Negativo) |
|--------------------------------------|--------------------------------|------------------------------------|
| **Modelo dice: "Es FRAUDE" (Positivo)** | Verdadero Positivo (VP) | Falso Positivo (FP) |
| **Modelo dice: "NO es Fraude" (Negativo)** | Falso Negativo (FN) | Verdadero Negativo (VN) |

## Verdadero Positivo (VP)

El modelo dijo que **sí era fraude** y efectivamente **sí era fraude**.

✅ Es un **acierto**.

## Verdadero Negativo (VN)

El modelo dijo que **no era fraude** y efectivamente **no era fraude**.

✅ Es un **acierto**.

## Falso Positivo (FP) — Error Tipo 1

El modelo dijo que **sí era fraude**, pero en realidad **no lo era**.

🚨 Es una **falsa alarma**.

## Falso Negativo (FN) — Error Tipo 2

El modelo dijo que **no era fraude**, pero en realidad **sí era fraude**.

⚠️ Es un **peligro no detectado**.

---

# 3. ¿Cómo afecta mover el umbral a estos valores?

Existe un **equilibrio entre los Falsos Positivos (FP) y los Falsos Negativos (FN)** dependiendo de dónde coloquemos el umbral.

## Si bajamos el umbral al 20%

El modelo se vuelve más **desconfiado**.

Ante la menor duda, marcará algo como **Positivo**.

### Resultado:

- 📈 Aumentan los **Verdaderos Positivos (VP)**.
- 📈 También aumentan los **Falsos Positivos (FP)**.
- Se detectan más casos positivos reales, pero también habrá más **falsas alarmas**.

### Ejemplo: Medicina

En la detección de **cáncer**, puede ser conveniente utilizar un umbral bajo para intentar no dejar ningún enfermo sin detectar, aunque esto implique realizar pruebas adicionales a algunas personas sanas.

---

## Si subimos el umbral al 80%

El modelo se vuelve más **prudente**.

Solo marcará algo como **Positivo** cuando tenga una probabilidad bastante alta de que realmente lo sea.

### Resultado:

- 📉 Disminuyen los **Falsos Positivos (FP)**.
- 📈 Aumentan los **Falsos Negativos (FN)**.
- Hay menos falsas alarmas, pero pueden escaparse algunos casos positivos reales.

### Ejemplo: Filtro de spam

En un filtro de **spam agresivo**, no queremos que correos importantes de trabajo terminen en la carpeta de correo no deseado.

Por eso, podemos utilizar un umbral más alto para que el modelo solo marque como spam los correos de los que esté muy seguro.
